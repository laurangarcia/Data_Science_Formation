Un sistema lineal se expresa como:   A * x = b

A : Matriz de coeficientes (m *n)

x : Vector de incognitas (n * 1)

b : Vector de terminos independientes (m * 1)



## Tipos de sistemas


1. Sistema con solución única

    

     * 𝐴 es cuadrada ( 𝑛 × 𝑛). 
     * La determinante es diferente de 0
     * Geométricamente: las rectas/planos se cortan en un solo punto.

2. Sistema sin solución

    

    * Inconsistente.
    * Geométricamente: rectas/planos paralelos que nunca se cruzan.

3. Sistema con infinitas soluciones

    

    * Las ecuaciones son linealmente dependientes.
    * Geométricamente: rectas/planos coinciden.

## Tipos de Resolución

Método de la matriz inversa (teórico, no recomendado en práctica): x = A¨-1 *b

```python
import numpy as np

A = np.array([[2, 1],
              [1, -3]])
b = np.array([5, -4])

x = np.dot(np.linalg.inv(A), b)
print("Solución con inversa:", x)
```

```python
#mostrar resultados que están muy próximos a 0
np.set_printoptions(suppress = True)
A = np.array([[3, 1], [2, 1]])
print(A)
```

```python
b = np.array([[1],[1]])
Ainversa = np.linalg.inv(A)
print(Ainversa)
```

```python
x = Ainversa.dot(b)
print(x)
```

```python
print(A.dot(x))
```

Método directo con np.linalg.solve

```python
x = np.linalg.solve(A, b)
print("Solución con solve:", x)
```

Eliminación Gaussiana, Basado en operaciones elementales que transforman 
𝐴 en una matriz triangular, resolviendo por sustitución regresiva.

1. Transformar la matriz a forma triangular superior.
2. Resolver de abajo hacia arriba.

Este método es la base de: Descomposición LU 

Sistemas sobredeterminados y mínimos cuadrados
Cuando 𝑚 > 𝑛 (más ecuaciones que incógnitas), generalmente no hay solución exacta.
Buscamos la solución que minimiza el error cuadrático:

```python
x, residuals, rank, s = np.linalg.lstsq(A, b, rcond=None)
```

## Sistema de ecuaciones lineales sin solución

```python
import matplotlib.pyplot as plt

x = np.arange(-6, 6) # Definimos nuestro rango


y1 = 3 * x + 5
y2 = -1 * x + 3
y3 = 2 * x + 1

plt.figure()
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xlim(-8,8)
plt.ylim(-8,8)

plt.axvline(x = 0, color =  "grey")
plt.axhline(y = 0, color =  "grey")
plt.show()
```

Este es un sistema sobredeterminado, tiene más ecuaciones, que variables, entonces no tiene solución

## Sistema con una solución

```python
x = np.arange(-6, 6)

y2 = -1 * x + 3
y3 = 2 * x + 1

plt.figure()
plt.plot(x, y1)
plt.plot(x, y2)

plt.xlim(-8,8)
plt.ylim(-8,8)

plt.axvline(x = 0, color =  "grey")
plt.axhline(y = 0, color =  "grey")
plt.show()
```
