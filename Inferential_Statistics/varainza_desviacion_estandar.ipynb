```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## We read the database
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
names = ["sepal_length", "sepal-width", "petal-length", "petal-width", "class"]
iris = pd.read_csv(url, names = names)
iris
```

```python
plt.hist(iris["sepal_length"], color = "orange", bins = int(180/5))
```

```python
sns.distplot(iris["sepal_length"], hist = True, kde = True, 
             bins = int(180/5), color = "orange")
```

## Variance Calculation

```python
iris["sepal_length"].var()
```

```python
iris["sepal_length"].std()
```

```python
iris["sepal_length"].mean()
```

### Creating a sample using simple random sampling

```python
sample = iris.sample(frac = 0.5)
sample
```

### Variance and Standar Deviation calculation with the sample

```python
sample["sepal_length"].var()
```

```python
sample["sepal_length"].std()
```

```python
sample["sepal_length"].mean()
```
